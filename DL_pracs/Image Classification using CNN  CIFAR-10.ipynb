```python
# 1. Import packages (same as the reference notebook)
import tensorflow as tf
from tensorflow import keras
from keras.datasets import cifar10
from keras import layers
import matplotlib.pyplot as plt
import numpy as np
```

```python
# 2. Load CIFAR-10 dataset - use ONE option

# OPTION 1: Directly from Keras (ACTIVE, same as reference)
# First run may download CIFAR-10; future runs use the Keras cache.
(train_x, train_y), (test_x, test_y) = cifar10.load_data()

# OPTION 2: CIFAR-10 .npz file in the same folder (OFFLINE)
# Comment Option 1 and uncomment all lines below.
# data = np.load("cifar10.npz")
# train_x, train_y = data["x_train"], data["y_train"]
# test_x, test_y = data["x_test"], data["y_test"]

# OPTION 3: CIFAR-10 .csv files in the same folder (OFFLINE)
# Each CSV: first column = label, next 3072 columns = pixel values.
# Comment Option 1 and uncomment all lines below.
# import pandas as pd
# train = pd.read_csv("cifar10_train.csv")  # add header=None if no header row
# test = pd.read_csv("cifar10_test.csv")    # add header=None if no header row
# train_y = train.iloc[:, 0].to_numpy()
# train_x = train.iloc[:, 1:].to_numpy().reshape(-1, 32, 32, 3)
# test_y = test.iloc[:, 0].to_numpy()
# test_x = test.iloc[:, 1:].to_numpy().reshape(-1, 32, 32, 3)
```

```python
# Step 3: Display dataset shapes
print("X_train:", train_x.shape)
print("Y_train:", train_y.shape)
print("X_test:", test_x.shape)
print("Y_test:", test_y.shape)
```

```python
# Step 4: Display one image of each class (0 to 9)
plt.figure(figsize=(12, 3))
for digit in range(10):
    index = np.where(train_y == digit)[0][0]
    plt.subplot(1, 10, digit + 1)
    plt.imshow(train_x[index])
    plt.title(str(digit))
    plt.axis("off")
plt.tight_layout()
plt.show()
```

```python
# Step 5: Normalize images
train_x = train_x / 255.0
test_x = test_x / 255.0
```

```python
# Step 6: Build CNN model
model = keras.Sequential([
    layers.Input(shape=(32, 32, 3)),
    layers.Conv2D(32, (3, 3), activation="relu"),
    layers.MaxPooling2D((2, 2)),
    layers.Flatten(),
    layers.Dense(10, activation="softmax")
])
```

```python
# Step 7: Compile model
model.compile(optimizer="adam", loss="sparse_categorical_crossentropy", metrics=["accuracy"])
```

```python
# Step 8: Train model
history = model.fit(train_x, train_y, epochs=3, validation_split=0.1)
```

```python
# Step 9: Evaluate accuracy
loss, accuracy = model.evaluate(test_x, test_y)
print("Test Accuracy:", accuracy)
```

```python
# Step 10: Predict and display one image
prediction = model.predict(test_x[:1], verbose=0)
predicted_digit = np.argmax(prediction)
print("Actual Digit:", test_y[0])
print("Predicted Digit:", predicted_digit)
plt.imshow(test_x[0])
plt.title(f"Predicted: {predicted_digit}")
plt.show()
```

```python
# Step 11: Plot training and validation accuracy
plt.plot(history.history["accuracy"], label="Training")
plt.plot(history.history["val_accuracy"], label="Validation")
plt.title("CNN Accuracy")
plt.legend()
plt.show()
```
